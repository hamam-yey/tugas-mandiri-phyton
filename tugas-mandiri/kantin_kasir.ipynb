```python
#kasir_kantin.py
#Program: Smart Kasir Kantin Sekolah (Fase 1)
#Materi: Pertemuan 1-8

def input_angka_positif(pesan_input):
    """
    Mengambil input angka dari user dan menangani error jika salah ketik (try-except)
    hanya mnerima angka lebih dari 0.
    """
    try:
        nilai = float(input(pesan_input))
        if nilai <= 0:
            print("Nilai harus lebih besar dari 0!\n")
            return None
        return nilai
    except ValueError:
        print("Masukan angka harus berupa angka yang valid\n")
        return None

def hitung_diskon(total_subtotal, status_member, kode_voucher,):
    """
    menghitung diskon member berdasarkan subtotal belanja.
    -Santri/Member belanja >= 50.000 -> Diskon 10%  
    -Santri/Member belanja >= 25.000 -> Diskon 5%  
    -Non-member / belanja dibawah 25.000 -> tidak ada diskon
    -Kode voucher (HEMAT5K) -> Diskon 5.000
    """  

    diskon_member = 0
    diskon_voucher = 0


    if status_member.lower() == "ya":
        if total_subtotal >= 50000:
            diskon_member = total_subtotal * 0.10
        elif total_subtotal >= 25000:
            diskon_member = total_subtotal * 0.05

    if kode_voucher == "HEMAT5K":
        diskon_voucher = 5000

    return diskon_member + diskon_voucher
    


def cetak_struk(nama_barang,  qty,  harga, subtotal, nama_barang_2, qty_2, harga_2, subtotal_2, diskon, kode_voucher, infaq, total_akhir, bayar, kembalian,):
    """
    Mencetak struk belanja dengan layout rapi menggunakan f-string.
    """
    print("\n" + "=" * 40)
    print("     KANTIN HSI BOARDING SCHOOL    ")
    print("=" * 40)
    print(f"{nama_barang}({qty} @ Rp {harga}: Rp {subtotal})")
    print(f"{nama_barang_2}({qty_2} @ Rp {harga_2}: Rp {subtotal_2})")
    print(f"Diskon            : {diskon:,.0f}")
    print(f"Voucher           : {kode_voucher}")
    print(f"Infaq Masjid      : {infaq:,.0f}")
    print("=" * 40)
    print(f"TOTAL AKHIR       : {total_akhir:,.0f}")
    print(f"Tunai             : {bayar:,.0f}")
    print(f"kembalian         : {kembalian:,.0f}")
    print("=" * 40)
    print(f"Syukron Wa Jazaakumullahu Khayran")
   
def main():
    print("=== PROGRAM KASIR KANTIN SEKOLAH ===")

    # 1. Sesi input data barang
    nama_barang = input("Nama barang            : ")

    qty = input_angka_positif("Jumlah yang dibeli           : ")
    if qty is None:
        return
    
    harga = input_angka_positif("Harga Satuan (Rp)      ")
    if harga is None:
        return

    nama_barang_2 = input("Nama barang            : ")

    qty_2 = input_angka_positif("Jumlah yang dibeli           : ")
    if qty_2 is None:
        return

    harga_2 = input_angka_positif("Harga Satuan (Rp)      ")
    if harga_2 is None:
        return

    status_member = input("Apakah santri/member? (ya/tidak): ")

    kode_voucher = input("Masukkan voucher jika punya")

    infaq = input("Ingin berinfaq sukarela? (ya/tidak):")
    if infaq == "ya":
        infaq = input_angka_positif("Nominal Infaq (Rp)")
        if infaq is None:
            return
    else:
        infaq = 0    
    

    # 2. Sesi Kalkulasi
    subtotal = qty * harga
    subtotal_2 = qty_2 * harga_2
    total_subtotal = subtotal + subtotal_2
    diskon = hitung_diskon(total_subtotal, status_member, kode_voucher,)
    total_akhir = subtotal + subtotal_2 - diskon + infaq

    print(f"\n>> Total tagihan: Rp {total_akhir:,.0f}")

    # 3. Sesi pembayaran
    bayar = input_angka_positif("Jumlah Uang dibayarkan (Rp) : ")
    if bayar is None:
        return

    # 4. Validasi Pembayaran
    if bayar < total_akhir:
        kurang = total_akhir - bayar
        print(f"Transaksi Gagal! Uang pembayaran kurang Rp {kurang:,.0f}")
    else:
        kembalian = bayar - total_akhir
        cetak_struk(nama_barang,  qty,  harga, subtotal, nama_barang_2, qty_2, harga_2, subtotal_2, diskon, kode_voucher, infaq, total_akhir, bayar, kembalian)


if __name__ == "__main__":
    main()
```
