```python
#Project mandiri kantin kasir (fase 2)

def input_angka_positif(pesan_input):
    """
    Mengambil input angka dari user dan menangani error jika salah ketik (try-except).
    Hanya menerima angka lebih dari 0. Mengembalikan None jika tidak valid.
    """
    try:
        nilai = float(input(pesan_input))
        if nilai <= 0:
            print("Nilai harus lebih besar dari 0!\n")
            return None
        return nilai
    except ValueError:
        print("Masukan harus berupa angka yang valid\n")
        return None


def hitung_diskon(total_subtotal, status_member, kode_voucher):
    """
    Menghitung total diskon berdasarkan subtotal belanja.
    - Santri/Member belanja >= 50.000 -> diskon 10%
    - Santri/Member belanja >= 25.000 -> diskon 5%
    - Non-member / belanja di bawah 25.000 -> tidak ada diskon member
    - Kode voucher HEMAT5K -> diskon Rp 5.000
    """
    diskon_member = 0
    diskon_voucher = 0

    if status_member.strip().lower() == "ya":
        if total_subtotal >= 50000:
            diskon_member = total_subtotal * 0.10
        elif total_subtotal >= 25000:
            diskon_member = total_subtotal * 0.05

    if kode_voucher == "HEMAT5K":
        diskon_voucher = 5000
    elif kode_voucher != "":
        print("Kode voucher tidak valid, diabaikan.")

    return diskon_member + diskon_voucher


def input_banyak_barang():
    """
    Mengumpulkan daftar barang satu pembeli.
    - Enter kosong  = selesai
    - Ketik 'batal' = hapus barang terakhir yang baru ditambahkan (Task 4)
    """
    daftar_item = []
    while True:
        nama_mentah = input("\nNama barang (kosongkan dan enter untuk selesai, 'batal' untuk hapus barang terakhir): ")

        # Task 4: pembatalan satu barang (item terakhir)
        if nama_mentah.strip().lower() == "batal":
            if daftar_item:
                dibatalkan = daftar_item.pop()  # .pop() mengambil & menghapus item terakhir
                print(f"-> Dibatalkan: {dibatalkan[0]} x{dibatalkan[1]:.0f} (Rp {dibatalkan[3]:,.0f})")
            else:
                print("-> Belum ada barang yang bisa dibatalkan.")
            continue

        nama_barang = " ".join(nama_mentah.split()).title()

        if nama_barang == "":
            break

        qty = input_angka_positif("Jumlah (qty): ")
        if qty is None:
            continue

        harga = input_angka_positif("Harga Satuan (Rp): ")
        if harga is None:
            continue

        subtotal = qty * harga
        daftar_item.append([nama_barang, qty, harga, subtotal])
        print(f"-> Ditambahkan: {nama_barang} x{qty:.0f} = Rp {subtotal:,.0f}")

    return daftar_item


def simpan_riwayat(nama_file, daftar_item, total_akhir):
    """Menambahkan satu transaksi ke akhir file riwayat (mode append)."""
    with open(nama_file, "a", encoding="utf-8") as f:
        f.write("=== Transaksi ===\n")
        for nama, qty, harga, subtotal in daftar_item:
            f.write(f"{nama} x{qty:.0f} @ Rp {harga:,.0f} = Rp {subtotal:,.0f}\n")
        f.write(f"Total Akhir: Rp {total_akhir:,.0f}\n\n")


def tampilkan_ringkasan_harian(nama_file):
    """
    Task 2: membaca file riwayat dan menghitung jumlah transaksi
    dari berapa kali penanda '=== Transaksi ===' muncul.
    """
    try:
        with open(nama_file, "r", encoding="utf-8") as f:
            isi = f.read()
    except FileNotFoundError:
        print("Belum ada transaksi yang tercatat. Terima kasih!")
        return

    jumlah_transaksi = isi.count("=== Transaksi ===")
    print(f"Hari ini ada {jumlah_transaksi} transaksi. Terima kasih!")


def tampilkan_barang_terlaris(daftar_nama_terjual):
    """
    Task 3: mencari barang yang paling sering dibeli dalam satu sesi.
    daftar_nama_terjual = list nama barang dari semua pembeli (boleh berulang).
    """
    if not daftar_nama_terjual:
        print("Belum ada barang terjual pada sesi ini.")
        return

    # hitung kemunculan tiap nama barang
    hitungan = {nama: daftar_nama_terjual.count(nama) for nama in set(daftar_nama_terjual)}
    terbanyak = max(hitungan.values())
    # list comprehension: semua barang dengan jumlah tertinggi (jaga-jaga seri)
    terlaris = sorted([nama for nama, jumlah in hitungan.items() if jumlah == terbanyak])

    print(f"Barang paling sering dibeli: {', '.join(terlaris)} ({terbanyak}x)")


def main():
    NAMA_FILE_RIWAYAT = "riwayat_transaksi.txt"
    barang_terjual_sesi = []  # Task 3: akumulasi nama barang dari semua pembeli

    print("=== PROGRAM KASIR KANTIN SEKOLAH (Fase 2) ===")
    # Loop luar melayani banyak pembeli
    while True:
        print("\n--- Pembeli Baru ---")
        daftar_item = input_banyak_barang()

        # Task 1: transaksi kosong -> pesan jelas, tidak memanggil simpan_riwayat
        if not daftar_item:
            print("Tidak ada barang yang diinput. Transaksi kosong tidak dicatat.")
        else:
            subtotal_total = 0
            for item in daftar_item:
                subtotal_total += item[3]  # index 3 = subtotal per barang

            status_member = input("Apakah santri/member? (ya/tidak): ")
            kode_voucher = input("Kode voucher (kosongkan jika tidak ada): ").strip().upper()

            diskon = hitung_diskon(subtotal_total, status_member, kode_voucher)
            total_akhir = max(subtotal_total - diskon, 0)

            print(f"\n >> Subtotal: Rp {subtotal_total:,.0f}")
            print(f" >> Diskon: Rp {diskon:,.0f}")
            print(f" >> Total Tagihan: Rp {total_akhir:,.0f}")
            bayar = input_angka_positif("Jumlah Uang Dibayarkan (Rp): ")

            if bayar is not None and bayar >= total_akhir:
                kembalian = bayar - total_akhir
                print(f"Kembalian: Rp {kembalian:,.0f}")
                simpan_riwayat(NAMA_FILE_RIWAYAT, daftar_item, total_akhir)
                print(f"Transaksi tersimpan di {NAMA_FILE_RIWAYAT}")

                # Task 3: catat nama barang hanya jika transaksi benar-benar terjual
                for item in daftar_item:
                    barang_terjual_sesi.append(item[0])  # index 0 = nama barang
            else:
                print("Transaksi dibatalkan (pembayaran kurang/tidak valid).")

        lagi = input("\nLayani pembeli lagi (ya/tidak): ")
        if lagi.strip().lower() != "ya":
            print("\n=== Ringkasan ===")
            tampilkan_ringkasan_harian(NAMA_FILE_RIWAYAT)  # Task 2: tepat sebelum break
            tampilkan_barang_terlaris(barang_terjual_sesi)  # Task 3
            print("\nKasir ditutup. Syukron Jazaakumullahu Khayran")
            break


if __name__ == "__main__":
    main()
```
