```python
#LATIHAN 1
s = "Belajar Python itu Menyenangkan"
b = s[1]
n = s[-1]
spasi = [7]

print(f"kata asli     : {s}")
print(f"huruf pertama : {b}")
print(f"huruf terakhir: {n}")
print(f"Spasi         : {spasi}")
```

```python
#LATIHAN 2
s = "Belajar Python itu Menyenangkan"

kata_pertama = s[0:7:1]
kata_kedua = s[8:14:1]
kata_ketiga = s[19::1]

print(f"kata pertama    : {kata_pertama}")
print(f"kata kedua      : {kata_kedua}")
print(f"kata ketiga     : {kata_ketiga}")
```

```python
#LATIHAN 3

kalimat = input("Masukkan kata untuk dibalik!")#"mamah hamam"
kalimat_terbalik = kalimat[::-1]
if kalimat == kalimat_terbalik:
    print(f"kalimatnya dibalik sama aja btw = {kalimat_terbalik}")
else:
    print(f"'{kalimat}' dibalik menjadi '{kalimat_terbalik}'")
```

```python
#LATIHAN 4

kata_rahasia = "BPrOaGikRtoAdMiItMuINtuGhpPyYcThHoOnN"
isi_kata_rahasia = kata_rahasia[::3]

print(f"isi kata rahasia = {isi_kata_rahasia}")
```

```python
#LATIHAN 5

nama_rusak = "mUHAMMAD fAHRI"
nama1 = nama_rusak[0:8:]
nama2 = nama_rusak[9:14:]
nama_benar1 = nama1[0].upper() + nama1[1:].lower()
nama_benar2 = nama2[0].upper() + nama2[1:].lower()
hasil = nama_benar1 + nama_benar2

print(f"Nama rusak   : {nama_rusak}")
print(f"Nama pertama : {nama_benar1}")
print(f"Nama kedua   : {nama_benar2}")
print(f"Nama benar   : {hasil}")
```
