```python
#Latihan 1
biodata = {
    "nama": "Budi Santoso",
    "umur": 17,
    "hobi": ["membaca", "sepak bola", "memancing"],
    "sudah_menikah": False
}

# Cetak seluruh dictionary
print(biodata)

# Cetak value dari key "nama"
print(biodata["nama"])

# Cetak hobi pertama
print(biodata["hobi"][0])
```

```python
#Latihan 2
biodata["kota"] = "Bandung"

# Ubah umur menjadi umur tahun depan
biodata["umur"] = 18

# Cetak dictionary yang sudah diperbarui
print(biodata)
```

```python
#Latihan 3

# Menggunakan .get() tanpa nilai default (hasilnya None)
print(biodata.get("pekerjaan"))

# Menggunakan .get() dengan nilai default "Pelajar"
print(biodata.get("pekerjaan", "Pelajar"))
```

```python
#Latihan 4 
kalimat = input("Masukkan sebuah kalimat: ").lower()
kata_kata = kalimat.split()

histogram = {}
for kata in kata_kata:
    histogram[kata] = histogram.get(kata, 0) + 1

print(histogram)
```
