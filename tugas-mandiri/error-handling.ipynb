```python
    #SOAL 1
try:
    angka_pertama = int(input("Masukkan angka pertama"))
    operator = input("Masukkan operator (+ x - :)")
    angka_kedua = int(input("Masukkan angka kedua"))

    if operator == "+":
        hasil = angka_pertama + angka_kedua
        print(f"hasilnya adalah : {hasil}")
    elif operator == "-":
        hasil = angka_pertama - angka_kedua
        print(f"hasilnya adalah : {hasil}")
    elif operator == "x":
        hasil = angka_pertama * angka_kedua
        print(f"hasilnya adalah : {hasil}")
    elif operator == ":":
        hasil = angka_pertama / angka_kedua
        print(f"hasilnya adalah : {hasil}")
except ValueError:
    print("Masukkan angka!")   

except ZeroDivisionError:
    print("Tidak bisa membagi dengan nol(0)!")   

```

```python
#SOAL 2
try:
    angka_pertama = float(input("Masukkan angka pertama"))
    angka_kedua = float(input("Masukkan angka kedua"))
    angka_ketiga = float(input("Masukkan angka ketiga"))

    rata_rata = (angka_pertama + angka_kedua + angka_ketiga) / 3
    print(f"Jumlah rata-ratanya adalah {rata_rata}")
except ValueError:
    print("Masukkan angka!")   


```

```python

#SOAL 3
try:
    celcius = float(input("Masukkan suhu celsius"))
    fahrenheit = (celcius * 9/5) + 32
except ValueError: 
    print("Masukkan angka!")
else:
    print(f"hasil konversi: {fahrenheit}")
finally:
    print("=== Konversi selesai ===")
```

```python
try:
    harga_barang = float(input("Masukkan harga barang"))
    if harga_barang <= 0:
        print("harga harus diatas nol!")
    else:
        diskon = float(input("Masukkan persen diskon"))
        if diskon < 0 or diskon > 100:
            print("diskon harus diantara 0-100%")
        else: 
            harga_setelah_diskon = harga_barang - (harga_barang * diskon / 100)
            jumlah_diskon = harga_barang * diskon / 100
            print(f"harga barang :{harga_barang:.2f}Rp ")
            print(f"jumlah_diskon :{jumlah_diskon:.2f}Rp ")
            print(f"harga setelah diskon: {harga_setelah_diskon:.2f} ")

except ValueError: 
    print("Masukkan angka!")
        
finally:
    print("=== Kalkulasi selesai ===")


```
