```python
#SOAL3

def analisis_angka(a, b, c):
    total_ketiga_angka = a + b + c
    rata2_ketiga_angka = (a + b + c) / 3
    terbesar_ketiga_angka = max(a, b, c)
    return total_ketiga_angka, rata2_ketiga_angka, terbesar_ketiga_angka

total_ketiga_angka, rata2_ketiga_angka, terbesar_ketiga_angka = analisis_angka(10, 20, 15)

print("total_ketiga_angka:",total_ketiga_angka)
print("rata2_ketiga_angka:",rata2_ketiga_angka)
print("terbesar_ketiga_angka:",terbesar_ketiga_angka)

    
```

```python
#SOAL4
def kalkulasi_persegi_panjang(panjang, lebar):
    """
    fungsi ini menerima ukuran panjang dan lebar dalam m lalu
    mengembalikan nilai luas dan keliling sebagai int. 

    Args: 
        panjang(float): Nilai panjang dalam meter
        lebar(float): Nilai lebar dalam meter
    Returns:
         tuple: Nilai luas (m²) dan keliling (m).

    """
    luas = panjang * lebar
    keliling = 2 * (panjang + lebar)
    return luas, keliling

print(kalkulasi_persegi_panjang.__doc__)
```
