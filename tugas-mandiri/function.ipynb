```python
#SOAL4
def hitung_harga_akhir(harga_barang, persen_pajak):
    jumlah_pajak = harga_barang * (persen_pajak / 100) 
    harga_akhir = harga_barang + jumlah_pajak
    return harga_akhir

harga_bengbeng = float(input("Masukkan harga barang"))
persen_pajak = float(input("Masukkan persen pajak"))

hasil_akhir = hitung_harga_akhir(harga_bengbeng, persen_pajak)
print(f"Total harga akhir: {hasil_akhir:.2f}Rp")
   
```

```python
def fahrenheit_ke_celsius(temp_f):
    return (temp_f - 32) * 5/9

fahrenheit = float(input("Masukkan suhu fahrenheit"))
hasil = fahrenheit_ke_celsius(fahrenheit)
print(f"Konversi F ke C : {hasil}")
```
