```python
nama_file = input("Masukkan nama file: ")
file_mbox = open(nama_file)

total = 0
count = 0

for baris in file_mbox:
    baris = baris.rstrip()
    if baris.startswith("X-DSPAM-Confidence:"):
        angka_teks = baris[20:]
        angka = float(angka_teks)
        total += angka
        count += 1

print("Rata-rata spam confidence:", total / count)
```
