```python
#Latihan 2

kalimat = "Python"
kalimat_terbalik = ""

for huruf in kalimat:
    kalimat_terbalik = huruf + kalimat_terbalik

print(kalimat_terbalik)
```

```python
#Latihan 3

jumlah_angka = 0

for i in range (1, 51):
    if i % 3 == 0 and i % 5 == 0:
        jumlah_angka += 1

print(f"Jumlah angka yang habis dibagi 3 dan 5: {jumlah_angka}")

```

```python
#Latihan 4

for i in range(5, 0, -1):
    for j in range (i):
        print("*", end="")
    print()

print(i)
```

```python
#Latihan 5

angka_input = int(input("Masukkan angka bulat positif"))
hasil_faktorial = 1

for angka in range(1, angka_input + 1, 1):
    hasil_faktorial = hasil_faktorial * angka 

print(hasil_faktorial)
```
