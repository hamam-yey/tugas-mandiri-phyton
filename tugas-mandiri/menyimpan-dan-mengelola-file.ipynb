```python
#SOAL 2
nama_file = "sumber.txt"


print("------ Baris Puisi Saat Ini ------")
try:
    with open(nama_file, 'r') as file:
        baris_puisi = file.readlines()
        if not baris_puisi:
            print("(tidak ada puisi)")
        else:
            for i, puisi in enumerate(baris_puisi, 1):
                print(f"{i}.{puisi.strip()}")
except FileNotFoundError:
     print("(File puisi belum ada, akan dibuat baru)")
     baris_puisi = []

print("-" * 35)

baris_puisi_baru = input("masukkan baris puisi baru")
if baris_puisi_baru:
    baris_puisi.append(baris_puisi_baru + '\n')
    with open(nama_file, 'w') as file:
        file.writelines(baris_puisi)
        print("Tugas baru berhasil disimpan!")
else:
    print("Tidak ada tugas baru yang ditambahkan.")
```

```python
#SOAL 3
import os

nama_file = "catatan_harian.txt"
if os.path.exists(nama_file):
    print(f"File '{nama_file}' ditemukan")
    hapus_file = input("apakah anda tetap ingin menghapus? (y/n)")
    if hapus_file == "y":
        os.remove(nama_file)
        print("File berhasil dihapus")
    else:
        print("Penghapusan dibatalkan")
else:
    print(f"File '{nama_file}' tidak ada, tidak ada yang bisa dihapus.")

```
