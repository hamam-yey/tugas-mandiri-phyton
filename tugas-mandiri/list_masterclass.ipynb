```python
#Soal 3
kalimat = input("Masukkan kalimat!")
daftar_kalimat = kalimat.split()
daftar_kalimat.sort(key=str.lower)
jumlah_kalimat = len(daftar_kalimat)

print(f"Jumlah kalimat = {jumlah_kalimat}")
print(f"Kalimat sesuai urutan = {daftar_kalimat}")

```

```python
#Soal 4
a = [1, 2, 3, 4, 5]
b = a
c = a.copy()

b[1] = 20
c[1] = 30

print(f"a = {a}")
print(f"b = {b}")
print(f"c = {c}")

#karena a dan b mengakses list yang sama = hasil printnya sama "1, 20, 3, 4, 5"
#karena membuat list baru(copy) maka hasilnya "1, 30, 3, 4, 5"
```

```python
#Soal 5
nilai_ujian = [45, 67, 88, 52, 91, 76, 60]
nilai_lulus = [n for n in nilai_ujian if n > 65]
predikat = ["lulus" if n >= 65 else "remedial" for n in nilai_ujian]


print(nilai_lulus)
print(nilai_ujian)
print(predikat)
```
