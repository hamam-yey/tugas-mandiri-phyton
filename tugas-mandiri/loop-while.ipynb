```python

#latihan 2

while True:
    try:
        tebakan_angka = int(input("Tebak angkanya!"))
        angka_rahasia = 7
        if tebakan_angka == angka_rahasia:
            print(f"Selamat, tebakan anda benar!)")
            break
        else:
            print(f"Angka salah, coba lagi!")
    except ValueError:
        print("masukinnya angka woy!")
print("Terima kasih sudah bermain!")
```

```python
#latihan 3
total = 0
count = 0

while True:
    try:
        angka_input = (input("Masukkan angka"))
        stopper = "done"
        if angka_input == stopper:
            print("Penghitungan selesai")
            break
        else:
            angka_insert = int(angka_input)
            total += angka_insert
            count += 1
    except ValueError:
        print("Input tidak Valid!")
        continue
if count > 0:
        rata2_angka = total / count
        print(f"Rata-rata angka: {rata2_angka}")
else:
    print("Belum ada angka yang dimasukkan")
print(f"Total:           {total}")
print(f"Banyaknya Angka: {count}")




```
