```python
#SOAL 1

jam_kerja_str = input("masukkan jam kerja")
jam_kerja = int(jam_kerja_str)
jam_kerja_normal = 40
tarif_str = input("masukkan tarif")
tarif = int(tarif_str)


print("mengecek jumlah tarif...")

if jam_kerja >= jam_kerja_normal:
    tarif_lembur = jam_kerja - 40
    tarif_normal = 40
    print("selamat anda mendapat tarif lembur")
else:
    tarif_normal = jam_kerja
    tarif_lembur = 0

upah = (tarif_normal * tarif) + (tarif_lembur * tarif * 1.5)

print(f"Jumlah upah anda adalah {upah}")
```

```python
#SOAL 3

insert_tinggi_badan = input("masukkan tinggi badan")
insert_berat_badan = input("masukkan berat badan")
try:

    tinggi_badan_cm = float(insert_tinggi_badan)
    berat_badan = float(insert_berat_badan)
except:
    print("ERROR, please enter a numeric inp")


tinggi_badan_m = tinggi_badan_cm / 100
bmi = berat_badan / (tinggi_badan_m ** 2)
    
if bmi >= 30:
    hasil = "Obesitas"
elif bmi >= 25:
    hasil = "gemuk"
elif bmi >= 18.5:
    hasil = "normal"
elif bmi <= 18.5 :
    hasil = "Kurus"
else:
    hasil = "ERROR"
print("BMI anda termasuk:", hasil)
```

```python
#SOAL 4

#Syarat masuk wahana = > 150 cm & > (12 tahun / didampingi orangtua)
insert_tinggi = input("Masukkan tinggi badan anda")
tinggi = float(insert_tinggi)
insert_usia = input("Masukkan usia anda")
usia = float(insert_usia)
insert_ada_pendamping = input("Apakah ada pendamping : (Iya/Tidak)")
didampingi_orangtua = (insert_ada_pendamping == "Iya")

tinggi_minimal = 150
usia_minimal = 12

if tinggi >= tinggi_minimal :
    print("Tinggi anda memenuhi syarat")
    if usia >= usia_minimal or didampingi_orangtua :
        print("Anda memenuhi syarat")    
    else:
        print("Anda belum cukup umur dan butuh pendamping")
else:
    print("Tinggi anda tidak memenuhi syarat")
```
