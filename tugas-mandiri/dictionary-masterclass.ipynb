```python
#Latihan 1
harga_buah = {"apel": 5000, "jeruk": 8500, "mangga": 7800, "pisang": 3000}

total = 0
for buah, harga in harga_buah.items():
    print(f"Harga 1 kg {buah} adalah Rp {harga}")
    total += harga

print(f"Total harga semua buah: Rp {total}")




```

```python
#Latihan 4
nama_file = "mbox-short.txt"
fhandle = open(nama_file)

hari_count = {}
for baris in fhandle:
    if baris.startswith("From "):
        kata_kata = baris.split()
        hari = kata_kata[2]
        hari_count[hari] = hari_count.get(hari, 0) + 1

fhandle.close()
print(hari_count)
```
